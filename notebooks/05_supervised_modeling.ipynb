# 05 Supervised Modeling
Spatial cross-validation for prospectivity classification and Au grade regression.

```python
import numpy as np
from sklearn.metrics import roc_auc_score
from aureo_ml.config import ELEMENTS
from aureo_ml.data.synthetic_generator import AntioquiaGeochemicalGenerator
from aureo_ml.evaluation.cross_validation import spatial_group_splits
from aureo_ml.models.classifier import build_classifier

df = AntioquiaGeochemicalGenerator().generate(3000)
features = ELEMENTS + ['depth_m','alteration_intensity','structural_score','lithology']
aucs = []
for train_idx, test_idx in spatial_group_splits(df, n_splits=5):
    model = build_classifier()
    model.fit(df.iloc[train_idx][features], df.iloc[train_idx]['prospective'])
    proba = model.predict_proba(df.iloc[test_idx][features])[:, 1]
    aucs.append(roc_auc_score(df.iloc[test_idx]['prospective'], proba))
np.mean(aucs), aucs
```

Expected output: spatial-CV AUC usually above 0.90 on synthetic data because targets are governed by realistic pathfinder controls.
