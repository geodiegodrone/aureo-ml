# 04 Clustering Analysis
K-Means, DBSCAN, and Gaussian Mixture Models for geochemical population detection.

```python
from aureo_ml.data.synthetic_generator import AntioquiaGeochemicalGenerator
from aureo_ml.models.clustering import cluster_geochemical_populations

df = AntioquiaGeochemicalGenerator().generate(2500)
clustered = cluster_geochemical_populations(df)
clustered.groupby('gmm_cluster')[['Au','As','Sb','Hg','Te']].median()
```

Expected output: one or more clusters show elevated Au-As-Sb-Hg-Te signatures compatible with hydrothermal enrichment.
