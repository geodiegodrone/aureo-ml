# 06 Explainability
SHAP global and local explanations, permutation importance, and PDP-ready model artifacts.

```python
import shap
from aureo_ml.config import ELEMENTS
from aureo_ml.data.synthetic_generator import AntioquiaGeochemicalGenerator
from aureo_ml.models.classifier import build_classifier

df = AntioquiaGeochemicalGenerator().generate(1200)
features = ELEMENTS + ['depth_m','alteration_intensity','structural_score','lithology']
model = build_classifier().fit(df[features], df['prospective'])
rf = model.named_steps['model']
x_trans = model.named_steps['preprocess'].transform(df[features].head(300))
explainer = shap.TreeExplainer(rf)
shap_values = explainer.shap_values(x_trans)
type(shap_values)
```

Expected output: SHAP values show Au, As, Sb, Hg, Te, alteration, and structural score as dominant controls.
