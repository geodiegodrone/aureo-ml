# 01 Data Generation
Generates 10,000 synthetic Antioquia geochemical samples with fixed seed and scientific controls.

```python
from pathlib import Path
from aureo_ml.data.synthetic_generator import AntioquiaGeochemicalGenerator

out = Path('../data/synthetic/antioquia_geochemistry.csv')
df = AntioquiaGeochemicalGenerator(random_seed=42).generate(10_000)
df.to_csv(out, index=False)
df.head()
```

Expected output: table with coordinates in Antioquia, 15 elements, Au in g/t, pathfinder enrichment, and prospectivity targets.
